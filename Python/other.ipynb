```python
import matplotlib.pyplot as plt
import numpy as np

def tau_A_mu_sig(decay,amplitude=None,center=None,sigma=None):
    if sigma is None:
        decay,amplitude,center,sigma= decay[0], decay[1], decay[2], decay[3]
    X = range(len(decay))

    pars_t = ['Lifetime','Amplitude','Center','FWHM']
    fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(2,2)

    axes = [ax1,ax2,ax3,ax4]
    for ax, tit,data in zip(axes,pars_t,pars):
        ax.set_title(tit)
        ax.get_xaxis().set_ticks([])
        ax.plot(X,data,'ko',markersize=4)
        ax.axhline(np.mean(data),color='k',alpha=0.4,zorder=-10,linestyle='dashed')

    for par,t in zip(pars,pars_t):
        if t == 'center':
            print('Average '+t+f' : {np.mean(par):.4g}')
        else:
            print('Average '+t+f' : {np.mean(par):.3g}')

    plt.tight_layout()
    plt.show()
        
```
