```python
# importing figure and show from bokeh.plotting
# module
from bokeh.plotting import figure,show
import bokeh.plotting as bb
# importing autompg from bokeh.sampledata.autompg
from bokeh.sampledata.autompg import autompg
from bokeh.palettes import Viridis256
# importing LinearColorMapper and ColorBar
# from bokeh.models
from bokeh.models import LinearColorMapper, ColorBar, Range1d

# importing transform from bokeh.transform
from bokeh.transform import transform
import pandas as pd
import numpy as np
from bokeh.transform import linear_cmap

df=pd.read_csv(r"C:\Users\jnich\OneDrive - USherbrooke\Uni\PhD\Data\micro-PL\26_05_08\Data_260508_1_1.txt",encoding='latin-1',comment='#',sep='\t',names=['x','detx','y','dety','count','count2','count1'])


x_vals = sorted(df['x'].unique())
y_vals = sorted(df['y'].unique())

mapper = LinearColorMapper(
    palette=Viridis256,
    low=df['count'].min(),
    high=df['count'].max()
)

p = figure(
    width=500, height=500,
    match_aspect=True,
    # x_range=(min(x_vals), max(x_vals)),
    y_range=(min(y_vals), max(y_vals))
)

grid = df.pivot(index='y', columns='x', values='count').values

p.image(
    image=[grid],
    x=min(x_vals), y=min(y_vals),
    dw=max(x_vals) - min(x_vals),
    dh=max(y_vals) - min(y_vals),
    color_mapper=mapper,
    # width=500, height=500,
)


colorbar = ColorBar(color_mapper=mapper, width=15, location=(0, 0),major_label_text_font_size='13pt')
# p.add_layout(colorbar, 'right')

p.xaxis.axis_label = "x (μm)"
p.yaxis.axis_label = "y (μm)"
p.x_range.range_padding = 0
p.xaxis.axis_label_text_font_size = '13pt'
p.yaxis.axis_label_text_font_size = '13pt'
p.xaxis.major_label_text_font_size = '13pt'
p.yaxis.major_label_text_font_size = '13pt'

show(p)
```
