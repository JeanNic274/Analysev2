```python
import numpy as np
from time import time
t =time()
def _boustrophedon_scan( index_ranges): # snake like order
    n = len(index_ranges)

    def recurse(level, parity):
        seq = list(index_ranges[level])
        if parity % 2 == 1:
            seq.reverse()

        if level == n - 1:
            for i in seq:
                yield (i,)
        else:
            for k, i in enumerate(seq):
                # parity accumulates, so every outer step flips the inner axes
                for rest in recurse(level + 1, parity + k):
                    yield (i,) + rest

    yield from recurse(0, 0)


axes = {'x':[0,10,1],'y':[0,10,1],'z':[0,10,1]}

axis_names = list(reversed(axes.keys()))
axis_values = []
for name in axis_names:
    start, stop, step = axes[name]
    if (start>stop and step >0) or (start<stop and step <0):
        start, stop = stop, start
    start, stop, step = 1000*int(start), 1000*int(stop), 1000*int(step)
    values = np.arange(start, stop + step / 2, step, dtype= int)
    axis_values.append(values)
    
index_ranges = [range(len(v)) for v in axis_values]
combos = list(_boustrophedon_scan(index_ranges))

for i, idx in enumerate(combos):
    coords = {
        name: axis_values[a][idx[a]]
        for a, name in enumerate(axis_names)
    }
    print(i,dict(list(coords.items())[-2:]))

print(time()-t)
print(axis_values)
```
